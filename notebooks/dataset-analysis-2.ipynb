```python
import pandas as pd
import numpy as np
```

```python
# Read only the header
cols = pd.read_csv(
    "/kaggle/input/datasets/joebeachcapital/windows-malwares/API_Functions.csv",
    nrows=0
).columns

print(len(cols))
```

```python
dtype_dict = {"SHA256": "string"}

for c in cols:
    if c != "SHA256":
        dtype_dict[c] = np.uint8

print("Done")
```

```python
api_df = pd.read_csv(
    "/kaggle/input/datasets/joebeachcapital/windows-malwares/API_Functions.csv",
    dtype=dtype_dict
)

api_df.info(memory_usage="deep")
```

```python
# DLL Dataset
dll_cols = pd.read_csv(
    "/kaggle/input/datasets/joebeachcapital/windows-malwares/DLLs_Imported.csv",
    nrows=0
).columns

dll_dtype = {"SHA256": "string"}

for c in dll_cols:
    if c != "SHA256":
        dll_dtype[c] = np.uint8

dll_df = pd.read_csv(
    "/kaggle/input/datasets/joebeachcapital/windows-malwares/DLLs_Imported.csv",
    dtype=dll_dtype
)

print("DLL Loaded")
dll_df.info(memory_usage="deep")
```

```python
# PE Header
header_cols = pd.read_csv(
    "/kaggle/input/datasets/joebeachcapital/windows-malwares/PE_Header.csv",
    nrows=0
).columns

header_dtype = {"SHA256": "string"}

for c in header_cols:
    if c != "SHA256":
        header_dtype[c] = np.int64

header_dtype["Type"] = np.uint8

pe_header_df = pd.read_csv(
    "/kaggle/input/datasets/joebeachcapital/windows-malwares/PE_Header.csv",
    dtype=header_dtype
)

print("PE Header Loaded")
pe_header_df.info(memory_usage="deep")
```

```python
# PE Section
section_cols = pd.read_csv(
    "/kaggle/input/datasets/joebeachcapital/windows-malwares/PE_Section.csv",
    nrows=0
).columns

section_dtype = {"SHA256": "string"}

for c in section_cols:
    if c != "SHA256":
        section_dtype[c] = np.int64

section_dtype["Type"] = np.uint8

pe_section_df = pd.read_csv(
    "/kaggle/input/datasets/joebeachcapital/windows-malwares/PE_Section.csv",
    dtype=section_dtype
)

print("PE Section Loaded")
pe_section_df.info(memory_usage="deep")
```

```python
print("API       :", api_df.shape)
print("DLL       :", dll_df.shape)
print("PE Header :", pe_header_df.shape)
print("PE Section:", pe_section_df.shape)
```

```python
api_sha = set(api_df["SHA256"])
dll_sha = set(dll_df["SHA256"])
header_sha = set(pe_header_df["SHA256"])
section_sha = set(pe_section_df["SHA256"])

common_sha = api_sha & dll_sha & header_sha & section_sha

print("API:", len(api_sha))
print("DLL:", len(dll_sha))
print("Header:", len(header_sha))
print("Section:", len(section_sha))
print("Common SHA256:", len(common_sha))
```

```python
# Remove duplicate Type columns before merging

dll_merge = dll_df.drop(columns=["Type"])

header_merge = pe_header_df.drop(columns=["Type"])

section_merge = pe_section_df.drop(columns=["Type"])
```

```python
master_df = (
    api_df
    .merge(dll_merge, on="SHA256", how="inner")
    .merge(header_merge, on="SHA256", how="inner")
    .merge(section_merge, on="SHA256", how="inner")
)

print(master_df.shape)
```

```python
master_df = master_df.drop_duplicates()

print(master_df.shape)
```

```python
for name, df in [
    ("API", api_df),
    ("DLL", dll_df),
    ("PE_Header", pe_header_df),
    ("PE_Section", pe_section_df)
]:
    print("="*40)
    print(name)
    print("Duplicate SHA:", df["SHA256"].duplicated().sum())
```

```python
dup_sha = master_df["SHA256"][master_df["SHA256"].duplicated(keep=False)]

print(dup_sha.value_counts())
```

```python
print("Before Cleaning")
print("----------------------------")
print("API :", api_df.shape)
print("DLL :", dll_df.shape)
print("Header :", pe_header_df.shape)
print("Section :", pe_section_df.shape)

api_df = api_df.drop_duplicates(subset="SHA256", keep="first")
dll_df = dll_df.drop_duplicates(subset="SHA256", keep="first")
pe_header_df = pe_header_df.drop_duplicates(subset="SHA256", keep="first")
pe_section_df = pe_section_df.drop_duplicates(subset="SHA256", keep="first")

print("\nAfter Cleaning")
print("----------------------------")
print("API :", api_df.shape)
print("DLL :", dll_df.shape)
print("Header :", pe_header_df.shape)
print("Section :", pe_section_df.shape)
```

```python
dll_merge = dll_df.drop(columns=["Type"])
header_merge = pe_header_df.drop(columns=["Type"])
section_merge = pe_section_df.drop(columns=["Type"])

master_df = (
    api_df
    .merge(dll_merge, on="SHA256", how="inner")
    .merge(header_merge, on="SHA256", how="inner")
    .merge(section_merge, on="SHA256", how="inner")
)

print(master_df.shape)
```
